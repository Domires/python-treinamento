# Jupyter Notebook

Um **notebook** (`.ipynb`) mistura **texto** (como este) e **código**, e guarda o **resultado** embaixo de cada célula.

**Antes de começar:** no canto superior direito, clique em **Select Kernel** → **Python Environments** → escolha o seu Python.

**Para rodar uma célula:** clique no ▶ ao lado dela ou aperte `Shift + Enter` (roda e vai para a próxima).

````python
print("Olá! Esta é a primeira célula de código.")
````

## Variáveis ficam guardadas

Rode a célula abaixo para criar as variáveis e depois a seguinte para usá-las.
Assim como no `.py` com `#%%`, as variáveis continuam existindo entre as células.

````python
planejado_kg = 20000
produzido_kg = 18500
````

````python
eficiencia = produzido_kg / planejado_kg
print(f"Eficiência: {eficiencia:.1%}")
````

## A última linha aparece sozinha

Não precisa de `print()` para ver o valor da **última linha** da célula.
Com o pandas, as tabelas aparecem formatadas:

````python
import pandas as pd

producao = pd.DataFrame({
    "unidade": ["Unidade_A", "Unidade_B", "Unidade_C"],
    "produzido_kg": [18500, 12100, 9800],
})
producao
````

## Células de texto (Markdown)

Esta célula é de **texto**. Dê um duplo clique nela para ver como ela é escrita:

- `#` no começo da linha vira **título**
- `**texto**` vira **negrito**
- `-` no começo da linha vira item de lista

Para voltar a ver o texto formatado, aperte `Shift + Enter`.

**Dica:** o botão **Restart** (no topo) apaga todas as variáveis e recomeça do zero. O **Run All** roda todas as células em ordem.

# Rodando Python - 04: Google Colab (no navegador)

O **Google Colab** é um Jupyter Notebook que roda nos computadores do Google.
Não precisa instalar nada: basta uma **conta Google** e um navegador.

## Começando

1. Acesse https://colab.research.google.com
2. Clique em **Novo notebook** (ou **Arquivo → Novo notebook**).
3. Escreva o código na célula e aperte `Shift + Enter` (ou clique no ▶ da célula).
4. Para criar células: botões **+ Código** e **+ Texto** no topo.

Na primeira execução, o Colab leva alguns segundos para "conectar" (ligar a máquina virtual do Google).

## Abrindo um notebook do treinamento

**Arquivo → Fazer upload de notebook** → escolha um arquivo `.ipynb` (por exemplo, `03_notebook.ipynb` desta pasta).

## Instalando bibliotecas

O pandas e o matplotlib já vêm instalados. Para outras bibliotecas, use `!pip` dentro de uma célula:

## Usando arquivos (planilhas, CSV...)

**Opção 1 — Upload rápido:** clique no ícone de **pasta** na barra lateral esquerda → **Fazer upload** → escolha o arquivo.
Depois, leia normalmente:

```python
import pandas as pd

producao = pd.read_excel("producao.xlsx")
```

> **Atenção:** os arquivos enviados assim são **apagados** quando a sessão termina (ao fechar a aba ou após um tempo parado).

**Opção 2 — Google Drive (os arquivos ficam salvos):**

```python
from google.colab import drive
drive.mount("/content/drive")

producao = pd.read_excel("/content/drive/MyDrive/treinamento/producao.xlsx")
```

Na primeira vez, o Colab pede permissão para acessar o seu Drive.

## Salvando o trabalho

- O notebook fica salvo automaticamente no seu Google Drive, na pasta **Colab Notebooks**.
- Para baixar: **Arquivo → Fazer download → Fazer download do .ipynb** (abre também no VS Code).
